```python
import numpy as np
import ase.io
from matplotlib import pyplot as plt
```

```python
def get_mae(first, second):
    delta = first - second
    return np.mean(np.abs(delta))

def get_norm_mae(first, second):
    norm_first = np.sqrt(np.sum(first * first, axis = 1))
    norm_second = np.sqrt(np.sum(second * second, axis = 1))
    return get_mae(norm_first, norm_second)
```

```python
predictions_energies, predictions_forces = [], []
for i in range(8):
    predictions_path = f'hme21_predictions_{i}/'
    predictions_energies_now = np.load(predictions_path + 'energies_predicted.npy')
    predictions_forces_now = np.load(predictions_path + 'forces_predicted.npy')
    
    '''structures = ase.io.read(f'../../../datasets/hme21/hme21_test_chunk_{i}.xyz', index = ':')
    forces = [struc.arrays['forces'] for struc in structures]
    forces = np.concatenate(forces, axis = 0)
    print(i, get_mae(forces, predictions_forces_now))'''
    predictions_energies.append(predictions_energies_now)
    predictions_forces.append(predictions_forces_now)

predictions_energies = np.concatenate(predictions_energies, axis = 0)
predictions_forces = np.concatenate(predictions_forces, axis = 0)

structures = ase.io.read('../../../datasets/hme21/hme21_test.xyz', index = ':')

energies = [struc.info['energy'] for struc in structures]
energies = np.array(energies)

forces = [struc.arrays['forces'] for struc in structures]
forces = np.concatenate(forces, axis = 0)

n_atoms = [len(struc.positions) for struc in structures]
n_atoms = np.array(n_atoms)

print(energies.shape, predictions_energies.shape)
print(forces.shape, predictions_forces.shape)


print('force components mae: ', get_norm_mae(forces, predictions_forces), 'eV/A')

energies_per_atom = np.array([energies[i] / n_atoms[i] for i in range(len(energies))])
predictions_energies_per_atom = np.array([predictions_energies[i] / n_atoms[i] for i in range(len(energies))])

#print(energies_per_atom.shape)
#print(predictions_energies_per_atom.shape)

print('energy mae per atom: ', get_mae(energies_per_atom, predictions_energies_per_atom), 'eV/atom')


```
