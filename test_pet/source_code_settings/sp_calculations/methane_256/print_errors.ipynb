```python
import ase.io
import numpy as np
```

```python
structures = ase.io.read('../../datasets/methane/methane_test.extxyz', index = ':')

energies = np.array([struc.info['energy'] for struc in structures])
forces = [struc.arrays['forces'] for struc in structures]
forces = np.concatenate(forces, axis = 0)

print(energies.shape)
print(forces.shape)
```

```python
def get_rmse(first, second):
    delta = first - second
    return np.sqrt(np.mean(delta ** 2))
```

```python
energies_predicted = np.load('100k_256_predictions/energies_predicted.npy')
print(get_rmse(energies, energies_predicted))

energies_predicted = np.load('300k_256_predictions/energies_predicted.npy')
print(get_rmse(energies, energies_predicted))
```
